Objetivo: identificar qué columnas de `create_tables.sql` se pueden tomar directamente del CSV, cuáles se derivan y cuáles se deben simular. El mapeo completo queda documentado en `diseno_modelo_relacional/mapeo_datos.md`.

### Dependencias funcionales

Se cuenta, para cada determinante, cuántos de sus grupos tienen más de un valor del dependiente. Cero violaciones indica que la dependencia se cumple en los datos.

```python
def fd_violations(df, lhs, rhs):
    """Cantidad de grupos de `lhs` con más de un valor de `rhs`."""
    return int((df.groupby(lhs)[rhs].nunique() > 1).sum())


dependencias = [
    ("Order ID", "Customer ID"),
    ("Order ID", "Order Date"),
    ("Order ID", "Ship Date"),
    ("Order ID", "Ship Mode"),
    ("Order ID", "Postal Code"),
    ("Customer ID", "Customer Name"),
    ("Customer ID", "Segment"),
    ("Customer ID", "Postal Code"),
    ("Customer ID", "Region"),
    ("Product ID", "Product Name"),
    ("Product ID", "Sub-Category"),
    ("Sub-Category", "Category"),
    ("State", "Region"),
    ("Postal Code", "City"),
    ("City", "State"),
]

pd.DataFrame(
    [(lhs, rhs, df[lhs].nunique(), fd_violations(df, lhs, rhs)) for lhs, rhs in dependencias],
    columns=["determinante", "dependiente", "grupos", "violaciones"],
)
```

- La orden determina cliente, fechas, modo de envío y dirección: la dirección es **de la orden**, no del cliente.
- `Customer ID → Postal Code` y `Customer ID → Region` fallan en casi todos los clientes. Como la regla 7 exige un único código postal por cliente, se asigna a cada cliente su dirección más frecuente.
- `Product ID → Product Name` falla: hay IDs reutilizados para productos distintos.
- `City → State` falla porque hay nombres de ciudad repetidos en varios estados; la ciudad se identifica por (ciudad, estado).

### Productos y precios

Precio de lista unitario: `Sales / Quantity / (1 - Discount)`. Costo unitario: `(Sales - Profit) / Quantity`.

```python
df["precio_venta"] = df["Sales"] / df["Quantity"] / (1 - df["Discount"])
df["precio_compra"] = (df["Sales"] - df["Profit"]) / df["Quantity"]

clave_producto = ["Product ID", "Product Name"]
productos = df.groupby(clave_producto)
rango = productos[["precio_venta", "precio_compra"]].agg(lambda s: s.max() - s.min())

print("Product ID distintos:", df["Product ID"].nunique())
print("Productos (ID, nombre):", productos.ngroups)
print("Productos con precio variable:", (rango > 0.01).sum().to_dict())
print("Precio mínimo:", round(df["precio_venta"].min(), 2), "| costo mínimo:", round(df["precio_compra"].min(), 2))
print("Líneas con costo unitario > precio de lista:", (df["precio_compra"] > df["precio_venta"]).sum())

largo = df["Product Name"].str.len()
print("Nombres de más de 64 caracteres:", (largo > 64).sum(), "| máximo:", largo.max())
```

Con la clave (ID, nombre) los precios son constantes, así que `precio_venta` y `precio_compra` se derivan sin ambigüedad y cumplen `> 0`. Los nombres exigen ampliar `nombre_prod` a `VARCHAR(128)`.

### Líneas repetidas en una misma orden

La PK de `detalle_compra` es (`id_compra`, `id_producto`).

```python
repetidas = df[df.duplicated(["Order ID", *clave_producto], keep=False)]

print("Líneas repetidas:", df.duplicated(["Order ID", *clave_producto]).sum())
print("Con descuento distinto:", fd_violations(repetidas, ["Order ID", *clave_producto], "Discount"))
repetidas.sort_values("Order ID")[["Order ID", "Product ID", "Quantity", "Discount", "precio_venta"]]
```

Tienen el mismo precio y descuento, así que se fusionan sumando la cantidad.

### Envíos

`Ship Mode` no tiene columna en el esquema; se usa para simular `envio.fecha_esperada`. `Ship Date` se toma como `fecha_entrega`.

```python
fecha_orden = pd.to_datetime(df["Order Date"], format="%m/%d/%Y")
fecha_envio = pd.to_datetime(df["Ship Date"], format="%m/%d/%Y")

print("Rango de fechas:", fecha_orden.min().date(), "a", fecha_orden.max().date())
(fecha_envio - fecha_orden).dt.days.groupby(df["Ship Mode"]).describe()[["min", "mean", "max"]]
```

Las ventas se reconstruyen exactamente; la utilidad difiere solo por el redondeo del costo a centavos.

### Resumen

| Origen | Tablas / columnas |
|---|---|
| **CSV directo** | `segmento`, `categoria`, `subcategoria`, `producto.nombre_prod`, nombre y apellido de `cliente_nat`, `detalle_compra.cantidad`, `detalle_compra.descuento_aplicado`, `envio.fecha_entrega` |

Todas las PK y FK usan los identificadores autoincrementales de la base de datos; los IDs del CSV solo sirven para resolver relaciones durante la carga.
